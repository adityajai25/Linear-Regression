```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
x = energy.iloc[:,0:8]
print("Independent variables : ",list(x.columns))
```

```python
y = energy.iloc[:,-1:-3:-1]
print("Dependent variables :",list(y.columns))
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y,train_size=0.8, random_state=0) #Splitting, 80% of dataset to train & 20% for testing
```

```python
x_train.count() , y_train.count(), x_test.count(), y_test.count() #614 data is used for training and 154 is used for testing
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
l_model=LinearRegression()
l_model.fit(x_train, y_train)
predictions=l_model.predict(x_test)
predictions
```

```python
# y_test and predictions are in pandas
y_test_df = pd.DataFrame(y_test, columns=['Heating_Load', 'Cooling_Load'])
predictions_df = pd.DataFrame(predictions, columns=['Heating_Load_Pred', 'Cooling_Load_Pred'])

# Creating the DataFrame for Heating Load
error_pred_heating = pd.DataFrame({
    'Actual_data': y_test_df['Heating_Load'],
    'Prediction_data': predictions_df['Heating_Load_Pred']
})
error_pred_heating['Error'] = error_pred_heating['Actual_data'] - error_pred_heating['Prediction_data']

# Creating the DataFrame for Cooling Load
error_pred_cooling = pd.DataFrame({
    'Actual_data': y_test_df['Cooling_Load'],
    'Prediction_data': predictions_df['Cooling_Load_Pred']
})
error_pred_cooling['Error'] = error_pred_cooling['Actual_data'] - error_pred_cooling['Prediction_data']

print("Heating Load Errors:")
print(error_pred_heating)
print("\nCooling Load Errors:")
print(error_pred_cooling)
```

```python
from sklearn.metrics import *

r2_on_test_without_vif=r2_score(y_test, predictions)
print(f"Accuracy score: {round(r2_on_test_without_vif*100,3)} without Variance Inflation Factor")
```

#### Lets use Variance Inflation Factor

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
vif_data = pd.DataFrame() #Creating an empty DF for vif
vif_data["feature"] = x.columns
vif_data["VIF"] = [variance_inflation_factor(x.values, i) for i in range(len(x.columns))]
print(vif_data) #It is used to detect multicollinearity
```

```python
#Dropping columns that has more VIF value to avoid overfitting.
energy = energy.drop(['Overall_height'],axis=1)
energy = energy.drop(['Relative_Compactness'],axis=1) 
energy
```

```python
x = energy.loc[:,['Surface_Area', 'Wall_area', 'Roof_area', 'Orientation', 'Glazing_area', 'Glazing_Area_Distribution']]
y = energy.iloc[:,-1:-3:-1]
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y,train_size=0.8, random_state=0)
x_train , y_train, x_test, y_test
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
l_model=LinearRegression()
l_model.fit(x_train, y_train)
predictions=l_model.predict(x_test)
```

```python
y_test_df = pd.DataFrame(y_test, columns=['Heating_Load', 'Cooling_Load'])
predictions_df = pd.DataFrame(predictions, columns=['Heating_Load_Pred', 'Cooling_Load_Pred'])

# Creating the DataFrame for Heating Load
error_pred_heating = pd.DataFrame({
    'Actual_data': y_test_df['Heating_Load'],
    'Prediction_data': predictions_df['Heating_Load_Pred']
})
error_pred_heating['Error'] = error_pred_heating['Actual_data'] - error_pred_heating['Prediction_data']

# Creating the DataFrame for Cooling Load
error_pred_cooling = pd.DataFrame({
    'Actual_data': y_test_df['Cooling_Load'],
    'Prediction_data': predictions_df['Cooling_Load_Pred']
})
error_pred_cooling['Error'] = error_pred_cooling['Actual_data'] - error_pred_cooling['Prediction_data']

print("Heating Load Errors:")
print(error_pred_heating)
print("\nCooling Load Errors:")
print(error_pred_cooling)
```

```python
from sklearn.metrics import *
 
r2_on_test_with_vif=r2_score(y_test, predictions)
print(f"Accuracy score: {round(r2_on_test_with_vif*100,3)} with Variance Inflation Factor")
```

```python
print(f"Accuracy score: {round(r2_on_test_with_vif*100,3)} with Variance Inflation Factor")
print(f"Accuracy score: {round(r2_on_test_without_vif*100,3)} without Variance Inflation Factor")
```
